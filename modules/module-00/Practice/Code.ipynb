# SESSION 3 — Namespaces, Call Stack & Scope 


```python
#Problem 9 — Local vs Global
x = 10

def test():
    x = 20
    print(x)

test() # output will be 20 as it is defined in local namespace within the scope of of function test()
print(x) # output will be 10 as it is defined outisde local namespace in the global namespace where x  = 10 
```

```python
# Problem 10 — Function Frame
x = 100

def outer():
    x = 200

    def inner():
        print(x)

    inner()

outer()
# inner() find x= 200 as it is called within the scope of outer() function 
```

```python
# Problem 11 — UnboundLocalError
x = 10

def test():
    print(x)
    x = 20

test() # Here unboundlocal error occured as in the local namespace of function test scope without assigning x  a value in local scope we used print(x)so no value is assigned in local namespace within the scope of test function
```

```python
# Problem 12 — global
x = 10

def change():
   global x
   x = 50
    

change()

print(x)  
```
