```python
# Accumulator Pattern
numbers = [7 ,4 ,9 ,2 ,8]
# We want 7 + 4 + 9 + 2 + 8 = 30
total = 0 

for current in numbers:
    total += current # total = total + current 

print(total)          

    
```

| `current` | `total` before | `total += current` | `total` after |
| --------: | -------------: | -----------------: | ------------: |
|         7 |              0 |              0 + 7 |             7 |
|         4 |              7 |              7 + 4 |            11 |
|         9 |             11 |             11 + 9 |            20 |
|         2 |             20 |             20 + 2 |            22 |
|         8 |             22 |             22 + 8 |            30 |


complexity : O(n)
Extra space : O(1)

# Index Based Reasoning

```python
numbers = [14, 7, 22 ,9 ,31]
index = 0
target = 9
for current in numbers:
    if current == target:
        print(index)
        break
    index = index + 1
    
            

        

    
```
